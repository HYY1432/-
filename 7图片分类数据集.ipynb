两个可视化数据集的函数

```python
import torch
from matplotlib import pyplot as plt
# d2l 是《动手学深度学习》封装好的库，通常需要 pip install d2l 来安装
# 如果在本地运行，也可以用标准的 matplotlib 代替 d2l.plt
# from d2l import torch as d2l 

def get_fashion_mnist_labels(labels):
    """返回Fashion-MNIST数据集的文本标签。"""
    text_labels = [
        't-shirt', 'trouser', 'pullover', 'dress', 'coat', 
        'sandal', 'shirt', 'sneaker', 'bag', 'ankle boot'
    ]
    return [text_labels[int(i)] for i in labels]

def show_images(imgs, num_rows, num_cols, titles=None, scale=1.5):
    """Plot a list of images."""
    # 计算画布大小
    figsize = (num_cols * scale, num_rows * scale)
    
    # 创建多行多列的子图网格
    # 注意：图片中 d2l.plt 实际就是 matplotlib.pyplot
    _, axes = plt.subplots(num_rows, num_cols, figsize=figsize)
    
    # 展平多维数组，方便后续遍历
    axes = axes.flatten()
    
    # 遍历每个子图和对应的图片
    for i, (ax, img) in enumerate(zip(axes, imgs)):
        if torch.is_tensor(img):
            # 图片张量（PyTorch Tensor）
            ax.imshow(img.numpy())
        else:
            # PIL图片
            ax.imshow(img)
            
        # 关闭坐标轴的刻度和边框，让图片更好看
        ax.axes.get_xaxis().set_visible(False)
        ax.axes.get_yaxis().set_visible(False)
        
        # 如果传入了标题，则设置标题
        if titles:
            ax.set_title(titles[i])
            
    return axes
```

定义load_data_fashion_mnist函数

```python
def load_data_fashion_mnist(batch_size, resize=None):
    """下载Fashion-MNIST数据集，然后将其加载到内存中。"""
    # 定义数据预处理方式：先把图片转换为PyTorch张量
    trans = [transforms.ToTensor()]
    
    # 如果传入了 resize 参数，就在 ToTensor 之前插入尺寸缩放操作
    if resize:
        trans.insert(0, transforms.Resize(resize))
        
    # 将多个预处理操作组合成一个流水线
    trans = transforms.Compose(trans)
    
    # 下载并加载训练集
    mnist_train = torchvision.datasets.FashionMNIST(
        root="../data",      # 数据存放的路径
        train=True,          # 指定为训练集
        transform=trans,     # 应用上面定义的预处理
        download=True        # 如果本地没有，则自动下载
    )
    
    # 下载并加载测试集（参数与训练集基本一致，但 train=False）
    mnist_test = torchvision.datasets.FashionMNIST(
        root="../data", 
        train=False,         # 指定为测试集
        transform=trans, 
        download=True
    )
    
    # 返回两个 DataLoader（数据加载器）
    return (
        # 训练集：打乱顺序 (shuffle=True)
        data.DataLoader(
            mnist_train, 
            batch_size, 
            shuffle=True, 
            num_workers=get_dataloader_workers()
        ),
        # 测试集：不需要打乱顺序 (shuffle=False)
        data.DataLoader(
            mnist_test, 
            batch_size, 
            shuffle=False, 
            num_workers=get_dataloader_workers()
        )
    )
```
