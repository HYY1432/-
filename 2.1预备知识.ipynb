多个张量结合一起,dim=0代表行方向拼接，dim=1代表列拼接

```python
X = torch.arange(12,dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0,1,4,3],[1,2,3,4],[4,3,2,1]])
torch.cat((X,Y),dim=0), torch.cat((X,Y),dim=1)
```

运行一些操作可能会导致为新结果分配内存

```python
before = id(Y)
Y = Y + X
id(Y) == before
```

执行原地操作

```python
z = torch.zeros_like(Y)
print('id(z):', id(z))
z[:] = X + Y
print('id(z):', id(z))
```

如果在后续计算中没有重复使用X，我们也可以使用X[:] = X+Y或X+=Y 来减少操作的内存开销

```python
before = id(X)
X += Y
id(X) == before
```
